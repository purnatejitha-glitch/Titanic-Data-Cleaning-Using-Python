```python
# Make a copy of the original dataset
df_clean = df.copy()

# Age: Fill missing values with median
df_clean['Age'] = df_clean['Age'].fillna(
    df_clean['Age'].median()
)

# Embarked: Fill missing values with mode
df_clean['Embarked'] = df_clean['Embarked'].fillna(
    df_clean['Embarked'].mode()[0]
)

# Cabin: Drop column if it has more than 50% missing values
if df_clean['Cabin'].isnull().mean() > 0.50:
    df_clean.drop(columns=['Cabin'], inplace=True)

print("Missing data handling completed!")

print("\nRemaining missing values:")
print(df_clean.isnull().sum())
```

Missing Data Handling – Justification

1. Age: Missing values were replaced using
   median imputation because the median is
   less sensitive to extreme values.

2. Embarked: Missing values were replaced
   using mode imputation because it is a
   categorical variable.

3. Cabin: The column was removed because
   it contains a high percentage of missing
   values, making reliable imputation difficult.

```python
print("Remaining missing values:")
print(df_clean.isnull().sum())
```

```python
# Store row count before cleaning
rows_before = len(df_clean)

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Store row count after cleaning
rows_after = len(df_clean)

# Calculate removed duplicates
duplicates_removed = rows_before - rows_after

print("Rows Before Cleaning:", rows_before)
print("Rows After Cleaning:", rows_after)
print("Duplicates Removed:", duplicates_removed)
```

```python
print("Sex Values:")
print(df_clean['Sex'].value_counts(dropna=False))

print("\nEmbarked Values:")
print(df_clean['Embarked'].value_counts(dropna=False))
```

```python
df_clean['Sex'] = (
    df_clean['Sex']
    .astype('string')
    .str.strip()
    .str.lower()
    .replace({
        'm': 'male',
        'f': 'female'
    })
    .str.capitalize()
)

print(df_clean['Sex'].value_counts())
```

```python
df_clean['Embarked'] = (
    df_clean['Embarked']
    .astype('string')
    .str.strip()
    .str.upper()
)

df_clean['Embarked'] = df_clean['Embarked'].replace({
    'SOUTHAMPTON': 'S',
    'CHERBOURG': 'C',
    'QUEENSTOWN': 'Q'
})

print(df_clean['Embarked'].value_counts())
```

## Step 7: Data Standardisation

### Objective

1. Removed leading and trailing spaces.

```python
numeric_columns = df_clean.select_dtypes(
    include=['int64', 'float64']
).columns

print("Numerical Columns:")
print(numeric_columns.tolist())
```

```python
outlier_report = []

for col in numeric_columns:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower_limit) |
        (df_clean[col] > upper_limit)
    ]

    outlier_report.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Limit': lower_limit,
        'Upper Limit': upper_limit,
        'Outlier Count': len(outliers)
    })

outlier_df = pd.DataFrame(outlier_report)

display(outlier_df.round(2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for col in numeric_columns:

    plt.figure(figsize=(7, 4))

    sns.boxplot(x=df_clean[col])

    plt.title(f'Outlier Detection - {col}')
    plt.xlabel(col)

    plt.show()
```

### Cleaning Decision

```python
# Function to generate data quality summary
def data_quality_summary(data):

    return {
        'Total Rows': len(data),
        'Total Columns': len(data.columns),
        'Missing Values': data.isnull().sum().sum(),
        'Duplicate Rows': data.duplicated().sum(),
        'Data Types': data.dtypes.astype(str).to_dict()
    }

# Before cleaning
before = data_quality_summary(df)

# After cleaning
after = data_quality_summary(df_clean)

# Create comparison table
comparison = pd.DataFrame({
    'Before Cleaning': [
        before['Total Rows'],
        before['Total Columns'],
        before['Missing Values'],
        before['Duplicate Rows']
    ],
    'After Cleaning': [
        after['Total Rows'],
        after['Total Columns'],
        after['Missing Values'],
        after['Duplicate Rows']
    ]
}, index=[
    'Total Rows',
    'Total Columns',
    'Missing Values',
    'Duplicate Rows'
])

display(comparison)
```
